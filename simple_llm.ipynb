```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()
```

```python
model = ChatGroq(model="openai/gpt-oss-20b")
```

```python
#create state

class LLMState(TypedDict):
    
    input_text: str
    output_text: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    #extract question from the state
    question = state['input_text']
    #form a prompt
    prompt = f"Question: {question}\nAnswer:"
    #askthe question to the llm model
    response = model.invoke(prompt).content

    #update the answer in the state
    state['output_text'] = response
    
    return state
```

```python
#create a state graph
workflow = StateGraph(LLMState)

#add nodes
workflow.add_node('llm_question', llm_qa)

#add edges
workflow.add_edge(START, 'llm_question')
workflow.add_edge('llm_question', END)

#compile the graph
flow=workflow.compile()
```

```python
# execute the compiled graph
initial_state = {"input_text": "What is the capital of France?"}

final_state = flow.invoke(initial_state)

print(final_state)
```
